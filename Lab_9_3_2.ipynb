Task-01 : Write python function to return sum of even and odd numbers in the given list.
• Incorporate manual docstring in code with Google Style

Prompt: Write a Python function named `sum_even_odd` that takes a list of numbers as input. The function should calculate and return the sum of the even numbers and the sum of the odd numbers in the list as a tuple. Include a Google-style docstring explaining the function's purpose, arguments, and return value.

```python
def sum_even_odd(numbers):
  even_num = 0
  odd_num = 0
  for num in numbers:
    if num % 2 == 0:
      even_num += num
    else:
      odd_num += num
  return even_num, odd_num

list = [1, 2, 3, 4, 5, 6, 7, 8, 9]
even_sum, odd_sum = sum_even_odd(list)
print(f"The sum of even numbers: {even_sum}")
print(f"The sum of odd numbers: {odd_sum}")
```

Explanation: This code defines a function sum_even_odd that takes a list of numbers as input. It initializes two variables, even_num and odd_num, to zero. It then iterates through each number in the input list. If a number is even (divisible by 2 with no remainder), it is added to even_num; otherwise, it is added to odd_num. Finally, the function returns a tuple containing the calculated sum of even numbers and the sum of odd numbers. The code then demonstrates how to use this function with an example list and prints the results.

 Task-02: Write python program for sru_student class with attributes like name, roll no.,
hostel_status and fee_update method and display_details method.
• Write comments manually for each line/code *block*

 Prompt: Write a Python program for an `sru_student` class. The class should have attributes for the student's `name`, `roll_no`, and `hostel_status`. Include a method called `fee_update` that changes a `fees_paid` attribute to `True`, and a method called `display_details` that prints the student's information. Add manual comments for each line or code block to explain the code.

```python

class sru_student():
  def __init__(self, name, roll_no, hostel_status):
    self.name = name
    self.roll_no = roll_no
    self.hostel_status = hostel_status
    self.fees_paid = False
  def fee_update(self):
    self.fees_paid = True
  def display_details(self):
    print(f"Name: {self.name}")
    print(f"Roll No.: {self.roll_no}")
    print(f"Hostel Status: {self.hostel_status}")
    print(f"Fees Paid: {self.fees_paid}")
student1 = sru_student("Alice Smith", "SRU456", True)

print("Initial details:")
student1.display_details()

student1.fee_update()

print("\nDetails after fee update:")
student1.display_details()
```

Explaination: This code defines a Python class named sru_student. The __init__ method is used to create new student objects with a name, roll number, and hostel status, and it sets the initial fees paid status to False. The fee_update method allows you to change the fees paid status to True. The display_details method prints out the student's information, including their name, roll number, hostel status, and whether their fees are paid.

Prompt: Write a Python script with 3–4 functions (e.g., calculator: add, subtract, multiply, divide). Incorporate manual docstring in code with NumPy Style.

```python
def add(a, b):
  return a + b

def subtract(a, b):
  return a - b

def multiply(a, b):
  return a * b

def divide(a, b):
  if b == 0:
    raise ZeroDivisionError("Division by zero is not allowed")
  return a / b
num1 = 7
num2 = 2
print(f"The addition: {add(num1, num2)}")
print(f"The subtraction: {subtract(num1, num2)}")
print(f"The multipication: {multiply(num1, num2)}")

try:
    print(f"The division: {divide(num1, num2)}")
    print(f"{num1} / 0 = {divide(num1, 0)}")
except ZeroDivisionError as e:
    print(e)
```

Explaination: This Python code defines a simple calculator with four functions: add, subtract, multiply, and divide. The add function takes two numbers and returns their sum, subtract returns their difference, and multiply returns their product. The divide function performs division but includes error handling to prevent division by zero, raising a ZeroDivisionError if the second argument is 0. The code then demonstrates the use of these functions with example values and includes a try-except block to show how the division function handles the case where the divisor is zero.
